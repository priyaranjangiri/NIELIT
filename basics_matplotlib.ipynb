```python
import matplotlib.pyplot as plt
import numpy as np

line_style= dict(marker="v",markersize=10,markerfacecolor="black",markeredgecolor="red",linestyle="solid",linewidth=3,color="teal")
x=np.array([2021,2022,2023,2024])
y1=np.array([18,27,50,10])
y2=np.array([20,30,40,25])

plt.title("class size",fontsize=20,family="arial",fontweight="bold",color="black")
plt.xlabel("year",fontweight="bold",color="blue")
plt.ylabel("number of students",fontweight="bold",color="blue")

plt.xticks(x)
plt.plot(x,y1,**line_style)
plt.plot(x,y2,**line_style)
plt.grid()
plt.show()
```
