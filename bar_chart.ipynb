```python
import numpy as np
import matplotlib.pyplot as plt

catagories =np.array(["Grains","Fruit","Vegetables","Protien","Dairy","Sweets"])
values = np.array([15,25,20,10,25,15])

plt.title("Food Consumption")
plt.xlabel("Food Catagories")
plt.ylabel("gms per day")

plt.xticks(color="red")
plt.yticks(color="red")

plt.bar(catagories, values,color="skyblue")
plt.show()
```
