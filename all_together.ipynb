```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

df=pd.read_csv("orders.csv")
print(df.head())

type_count=df["Category"].value_counts(ascending=True)
plt.barh(type_count.index,type_count.values)
plt.xlabel("number of products")
plt.ylabel("categories")
plt.title("categories of products")

plt.tight_layout()
plt.show()
```
