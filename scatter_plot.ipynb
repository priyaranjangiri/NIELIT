```python
import numpy as np
import matplotlib.pyplot as plt

x1=np.array([0,1,1,2,3,4,2,5,6])
y1=np.array([50,70,75,89,90,95,80,100,100])

x2=np.array([0,1,1,2,3,4,2,5,6])
y2=np.array([52,73,76,82,86,91,80,97,90])

plt.scatter(x1,y1,color= 'skyblue',alpha=0.7,s=100,label="Class A")
plt.scatter(x2,y2,color= 'red',alpha=0.7,s=100,label="Class B")
plt.xlabel('Hours Studied')
plt.ylabel('Grades')

plt.title('Hours Studied vs Grades')
plt.legend()

plt.show()
```
