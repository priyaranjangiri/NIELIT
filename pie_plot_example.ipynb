```python
import numpy as np
import matplotlib.pyplot as plt

catagories=np.array(["old","middle-aged","young adults","teenagers"])
values=np.array([300,340,260,100])

colors=["yellow","green","blue","red"]

plt.pie(values,labels=catagories,autopct="%1.1f%%",colors=colors,explode=[0,0,0,0.1],shadow= True)
plt.title("Age distribution of daily customers")

plt.show()
```
